## 1. Clone đúng branch và cài môi trường

Cell này không dùng clone URL placeholder hoặc clone nhánh mặc định. Repo runtime có thay đổi local/diverged sẽ bị từ chối cập nhật tự động; không reset hoặc xóa file của bạn.


````python
def command(args, **kwargs):
    if not kwargs.get("capture_output"):
        # Pipe child output through Python so Colab displays live progress.
        with subprocess.Popen(args, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                              text=True, bufsize=1, **kwargs) as process:
            for line in process.stdout:
                print(line, end="", flush=True)
            status = process.wait()
        if status:
            raise subprocess.CalledProcessError(status, args)
        return subprocess.CompletedProcess(args, status)
    try:
        return subprocess.run(args, check=True, **kwargs)
    except subprocess.CalledProcessError as error:
        if error.stderr:
            print(error.stderr)
        raise

if CONFIG["SYNC_REPOSITORY"]:
    if not REPO.exists():
        command(["git", "clone", "--single-branch", "--branch", CONFIG["BRANCH"],
                 CONFIG["REPO_URL"], str(REPO)])
    else:
        if not (REPO / ".git").exists():
            raise RuntimeError(f"{REPO} exists but is not a Git checkout. Choose a new REPO_DIR.")
        dirty = subprocess.check_output(
            ["git", "-C", str(REPO), "status", "--porcelain", "--untracked-files=no"], text=True).strip()
        if dirty:
            raise RuntimeError("Runtime checkout has tracked local changes. Preserve them and choose a new REPO_DIR.")
        origin = subprocess.check_output(
            ["git", "-C", str(REPO), "remote", "get-url", "origin"], text=True).strip()
        normalize_url = lambda value: value.rstrip("/").removesuffix(".git")
        if normalize_url(origin) != normalize_url(CONFIG["REPO_URL"]):
            raise RuntimeError("REPO_DIR points to a different repository. Choose a new directory.")
        command(["git", "-C", str(REPO), "fetch", "origin", CONFIG["BRANCH"]])
        current = subprocess.check_output(
            ["git", "-C", str(REPO), "branch", "--show-current"], text=True).strip()
        if current != CONFIG["BRANCH"]:
            exists = subprocess.run(["git", "-C", str(REPO), "show-ref", "--verify", "--quiet",
                                     "refs/heads/" + CONFIG["BRANCH"]]).returncode == 0
            if exists:
                command(["git", "-C", str(REPO), "switch", CONFIG["BRANCH"]])
            else:
                command(["git", "-C", str(REPO), "switch", "-c", CONFIG["BRANCH"], "FETCH_HEAD"])
        command(["git", "-C", str(REPO), "merge", "--ff-only", "FETCH_HEAD"])

# Pin parser; existing checkout from a completed run can remain detached.
command(["git", "-C", str(REPO), "checkout", "--detach", PARSER_COMMIT])
actual_commit = subprocess.check_output(["git", "-C", str(REPO), "rev-parse", "HEAD"], text=True).strip()
assert actual_commit == PARSER_COMMIT
if not (REPO / "src/tools/run_m2_full.py").is_file():
    raise RuntimeError("Missing full runner. Check REPO_DIR and fetch the updated m3-data-split branch.")
os.chdir(REPO)
if CONFIG["INSTALL_DEPENDENCIES"]:
    command([sys.executable, "-m", "pip", "install", "-q", "-r", str(REPO / "src/requirements-m2.txt")])

ENV = os.environ.copy()
ENV["PYTHONPATH"] = str(REPO / "src") + os.pathsep + ENV.get("PYTHONPATH", "")
print("Python:", sys.version)
print("Commit:", subprocess.check_output(["git", "rev-parse", "HEAD"], text=True).strip())
````

## 2. Mount Drive và kiểm tra layout đầu vào

Mỗi phần tử của `CORPUS_DIRS` phải là thư mục chứa các publication, ví dụ `.../corpus` hoặc `.../work/raw`.

Hai layout được hỗ trợ:

```text
corpus/<paper>/paper.json + vN/metadata.json + *.tex/*.bib
work/raw/<paper>/paper_metadata.json + sources/vN/manifest.json + *.tex/*.bib
```

Không trỏ vào file ZIP, thư mục một version, thư mục `parsed` hoặc thư mục cha còn chứa nhiều archive. Không chạy đồng thời hai runtime ghi vào cùng `RUN_DIR`.


````python
if CONFIG["USE_DRIVE"]:
    try:
        from google.colab import drive
    except ImportError as error:
        raise RuntimeError("USE_DRIVE=True requires Colab. Set False for local execution.") from error
    print("Connecting Google Drive...", flush=True)
    drive.mount("/content/drive")
    print("Drive connected.", flush=True)

import shutil, zipfile, hashlib
PERSONAL.mkdir(parents=True, exist_ok=True)
archive = Path(ZIP_PATH)
if not archive.is_file():
    raise FileNotFoundError(f"Upload ZIP backup vào My Drive trước: {archive}")
extract_root = Path("/content/m3_recovery_input")
extract_root.mkdir(parents=True, exist_ok=True)
done = extract_root / "extraction_complete.json"
zip_sha256 = hashlib.file_digest(archive.open("rb"), "sha256").hexdigest()
if done.exists():
    if json.loads(done.read_text())["zip_sha256"] != zip_sha256:
        raise ValueError("ZIP đã thay đổi; dùng thư mục input và RUN_DIR mới")
    print("Reuse extracted corpus")
else:
    with zipfile.ZipFile(archive) as z:
        for member in z.infolist():
            target = (extract_root / member.filename).resolve()
            if not target.is_relative_to(extract_root.resolve()):
                raise ValueError("Unsafe ZIP path")
        required_space = sum(m.file_size for m in z.infolist())
        if shutil.disk_usage(extract_root).free < required_space + 1024**3:
            raise RuntimeError("Không đủ dung lượng local để giải nén ZIP")
        print("Extracting", len(z.infolist()), "files locally...", flush=True)
        z.extractall(extract_root)
    done.write_text(json.dumps({"zip_sha256": zip_sha256}))
# Copy original M1 tables; do not crawl/reconstruct metadata when backup exists.
m1 = PERSONAL / "m1_tables"
m1.mkdir(exist_ok=True)
for table in ["papers", "references"]:
    original = extract_root / "2606.00734-2606.05733/work/tables" / f"{table}.parquet"
    target = m1 / original.name
    if target.exists():
        if hashlib.file_digest(target.open("rb"), "sha256").hexdigest() != hashlib.file_digest(original.open("rb"), "sha256").hexdigest():
            raise ValueError(f"Existing M1 file differs: {target}")
    else:
        shutil.copy2(original, target)
print("M1 tables:", m1)
import shutil
print("Validating publication metadata (up to eight concurrent readers)...", flush=True)
inspection = (
    "import json,sys; from tools.run_m2_full import inventory; "
    "print(json.dumps(inventory(sys.argv[1:]),ensure_ascii=False))"
)
inventory_result = command(
    [sys.executable, "-c", inspection, *CONFIG["CORPUS_DIRS"]],
    cwd=REPO, env=ENV, capture_output=True, text=True,
)
INPUT = json.loads(inventory_result.stdout)
print("Input publications:", INPUT["publications"])
print("Declared versions:", INPUT["declared_versions"])
print("Examples:", [(p["arxiv_id"], p["versions"]) for p in INPUT["papers"][:3]])
ids = {p["arxiv_id"] for p in INPUT["papers"]}
expected_ids = {f"2606.{i:05d}" for i in range(734, 5734)}
if ids != expected_ids:
    raise ValueError(f"Corpus không đúng 5.000 ID: missing={len(expected_ids-ids)}, extra={len(ids-expected_ids)}")
existing_parent = RUN
while not existing_parent.exists():
    existing_parent = existing_parent.parent
print("Available output filesystem space (GiB):", round(shutil.disk_usage(existing_parent).free / 1024**3, 2))
````

## 3. Parse full → audit → xuất Parquet

Flow thực thi:

```text
metadata từng publication → từng version → gather compilation path
→ hierarchy + cleanup/math normalization
→ bibitem/.bib extraction → reference field union + aliases theo version
→ đổi cite → full-text dedup → parsed/<paper>.json
→ audit → exports/<lượt xuất mới>/tables/{elements,hierarchy,bibtex}/*.parquet
```

`LIMIT=0` xử lý toàn bộ input đã chọn. Runner dùng CLI subprocess để multiprocessing và timeout không phụ thuộc trạng thái kernel notebook.

JSON được ghi vào `RUN_DIR/parsed` để resume. Thay parser, input selection, timeout hoặc bib scope cần `RUN_DIR` mới. Input không được sửa. Các version empty/not_downloaded/no_root/timeout/parse_error và publication reference-conflict được báo riêng.

Nếu còn paper/source lỗi, trạng thái là `partial` và các cell thống kê/xuất dữ liệu vẫn chạy. Một code return 0 ở bước orchestration không có nghĩa toàn bộ 20.000 bài đã hoàn tất.


````python
args = [
    sys.executable, "-u", str(REPO / "src/tools/run_m2_full.py"),
    "--corpus-dirs", *CONFIG["CORPUS_DIRS"],
    "--run-dir", str(RUN),
    "--workers", str(CONFIG["WORKERS"]),
    "--timeout", str(CONFIG["TIMEOUT"]),
    "--bib-scope", CONFIG["BIB_SCOPE"],
    "--limit", str(CONFIG["LIMIT"]),
    "--shard-size", str(CONFIG["SHARD_SIZE"]),
    "--expected-publications", str(CONFIG["EXPECTED_PUBLICATIONS"]),
]
if not CONFIG["RETRY_INCOMPLETE"]:
    args.append("--no-retry-incomplete")
command(args, cwd=REPO, env=ENV)
RESULT = json.loads((RUN / "latest_result.json").read_text(encoding="utf-8"))
SUMMARY = RESULT["summary"]
METRICS = RESULT["metrics"]
print("\nRUN STATUS:", RESULT["status"])
print("Selected publications:", SUMMARY["selected_publications"])
print("Paper outputs:", SUMMARY["publications"])
print("Quarantined publications:", SUMMARY.get("quarantined_publications", 0))
print("Versions without paper output:", SUMMARY["versions_without_paper_output"])
````

````python
# Verify coverage/schema; reference-quality flags may still leave the run partial.
selected_inputs = INPUT["papers"][:CONFIG["LIMIT"]] if CONFIG["LIMIT"] else INPUT["papers"]
selected_by_id = {paper["arxiv_id"]: paper for paper in selected_inputs}
assert SUMMARY["publications"] == len(selected_inputs), "Missing publication JSON"
assert SUMMARY["versions_without_paper_output"] == 0
assert sum(SUMMARY["version_statuses"].values()) == sum(len(p["versions"]) for p in selected_inputs)
assert SUMMARY["totals"]["hierarchy_problems"] == 0
assert EXPORT and EXPORT["schema_and_count_checks"] == "passed"
assert EXPORT["papers_exported"] == len(selected_inputs)
assert len(list((RUN / "parsed").glob("*.json"))) == len(selected_inputs)
assert not json.loads((RUN / "paper_failures.json").read_text(encoding="utf-8"))
reported_ids = "2606.01068 2606.01093 2606.01180 2606.01867 2606.02653 2606.03207 2606.03277 2606.03445 2606.03697 2606.03932 2606.04100 2606.04330 2606.04482 2606.04540 2606.04809 2606.04879 2606.04951 2606.05612".split()
recovered = []
for arxiv_id in reported_ids:
    if arxiv_id not in selected_by_id:
        continue
    paper = selected_by_id[arxiv_id]
    data = json.loads((RUN / "parsed" / (paper["paper_key"] + ".json")).read_text(encoding="utf-8"))
    keys = {entry["citation_key"] for entry in data["bibtex"]}
    assert len(keys) == len(data["bibtex"]), arxiv_id
    assert data["stats"]["hierarchy_problems"] == 0, arxiv_id
    for conflict in data.get("reference_conflicts", []):
        assert conflict["citation_key"] not in keys, arxiv_id
        assert len(set(conflict["candidate_keys"])) >= 2, arxiv_id
        assert set(conflict["candidate_keys"]) <= keys, arxiv_id
        assert conflict["citation_key"] not in data["citation_aliases"].get(str(conflict["version"]), {}), arxiv_id
    recovered.append({"arxiv_id": arxiv_id, "versions": len(data["versions"]),
                      "parsed_versions": sum(v["status"] == "parsed" for v in data["versions"]),
                      "reference_conflicts": len(data.get("reference_conflicts", []))})
VERIFICATION = {
    "checks": "passed", "run_status": RESULT["status"],
    "publication_outputs": SUMMARY["publications"],
    "version_statuses": SUMMARY["version_statuses"],
    "version_records": sum(SUMMARY["version_statuses"].values()),
    "table_rows": EXPORT["table_rows"],
    "hierarchy_problems": SUMMARY["totals"]["hierarchy_problems"],
    "quality_flags": SUMMARY["quality_flags"],
    "unresolved_citation_pairs": SUMMARY["unresolved_citation_occurrences"],
    "previously_failing_papers_checked": len(recovered),
    "recovered_papers": recovered,
}
verification_path = RUN / "pipeline_verification.json"
verification_temp = verification_path.with_suffix(".tmp")
verification_temp.write_text(json.dumps(VERIFICATION, ensure_ascii=False, indent=2), encoding="utf-8")
verification_temp.replace(verification_path)
print("PIPELINE VERIFICATION: PASSED")
print(json.dumps({k: v for k, v in VERIFICATION.items() if k != "recovered_papers"}, ensure_ascii=False, indent=2))
display(pd.DataFrame(recovered))
print("Saved:", verification_path)
print("Coverage/schema checks passed; review partial-source/reference flags separately.")
````
